## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Remove duplicate rows

users = users.drop_duplicates()
flights = flights.drop_duplicates()
hotels = hotels.drop_duplicates()

# Convert date columns

flights['date'] = pd.to_datetime(flights['date'])
hotels['date'] = pd.to_datetime(hotels['date'])

# Merge datasets

travel_df = flights.merge(users,left_on='userCode',right_on='code',how='left')

travel_df = travel_df.merge(
    hotels,
    on=['travelCode','userCode'],
    how='left',
    suffixes=('_flight','_hotel')
)

travel_df.head()
````

### What all manipulations have you done and insights you found?

Several preprocessing steps were performed to prepare the datasets for analysis. Duplicate records were removed to improve data quality. Date columns were converted into datetime format to facilitate time-based analysis. The Users, Flights, and Hotels datasets were merged using the common user and travel identifiers to create a unified dataset containing customer demographics, flight information, and hotel booking details. This integrated dataset enables comprehensive exploratory analysis and supports future machine learning tasks such as regression, classification, and recommendation.

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
merged_df = flights.merge(
    users,
    left_on='userCode',
    right_on='code',
    how='left'
)

merged_df = merged_df.merge(
    hotels,
    on=['travelCode','userCode'],
    how='left'
)

numeric_df = merged_df.select_dtypes(include=['number'])

plt.figure(figsize=(14,10))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
````

##### 1. Why did you pick the specific chart?

A correlation heatmap summarizes relationships among all numerical variables simultaneously.

##### 2. What is/are the insight(s) found from the chart?

Hotel total cost is strongly correlated with hotel price and stay duration.
Flight price shows positive correlation with travel distance.
Some variables exhibit little or no correlation, indicating independent behavior.

These relationships help identify the most influential features for predictive modeling.
